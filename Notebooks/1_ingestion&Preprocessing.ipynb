```python
clinical = pd.read_csv(
    "TCGA-BRCA.clinical.tsv.gz",
    sep="\t",
    compression='gzip'
)

clinical.head()
```

```python
print(expr.shape)
expr.head()
```

```python
print(clinical.shape)
clinical.head()
```

```python
expr = expr.set_index(expr.columns[0])
```

```python
expr.head()
```

```python
expr.info()
```

```python
clinical.info()
```

```python
expr_t = expr.T
```

```python
expr_t.shape
```

```python
clinical['sample_type.samples'].value_counts()
```

```python
expr_t.index[:5]
```

```python
clinical['submitter_id'].head()
```

```python
expr_t['patient_id'] = expr_t.index.str.slice(0,12)
```

```python
expr_t['patient_id'].head()

```

```python
merged = expr_t.merge(
    clinical,
    left_on='patient_id',
    right_on='submitter_id'
)
```

```python
merged.shape
```

```python
filtered = merged[
    merged['sample_type.samples'].isin([
        'Primary Tumor',
        'Solid Tissue Normal'
    ])
]
```

```python
filtered['label'] = filtered['sample_type.samples'].map({
    'Primary Tumor': 1,
    'Solid Tissue Normal': 0
})
```

```python
filtered['label'].value_counts()
```

Removing the duplicates by using labels directly from expression sample IDs.

```python
merged['patient_id'].duplicated().sum()
```

```python
merged['patient_id'].value_counts().head()
```

```python
expr_t['sample_code'] = expr_t.index.str.slice(13,15)
```

```python
expr_t['sample_code'].value_counts()
```

```python
expr_t = expr_t[
    expr_t['sample_code'].isin(['01', '11'])
]
```

```python
expr_t['label'] = expr_t['sample_code'].map({
    '01': 1,
    '11': 0
})
```

```python
expr_t['label'].value_counts()
```

```python
expr_t.info()
```

```python
final_dataset = expr_t.drop(columns=['sample_code'])

final_dataset.to_csv(
    "brca_clean_expression_dataset.csv"
)
```
